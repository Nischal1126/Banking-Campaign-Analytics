```python
import pandas as pd

# Read all files with correct separator
df1 = pd.read_csv(r"C:\Users\nisch\Downloads\bank+marketing\bank\Original_data\bank-additional-full.csv", sep=';')
df2 = pd.read_csv(r"C:\Users\nisch\Downloads\bank+marketing\bank\Original_data\bank-additional.csv", sep=';')
df3 = pd.read_csv(r"C:\Users\nisch\Downloads\bank+marketing\bank\Original_data\bank-full.csv", sep=';')
df4 = pd.read_csv(r"C:\Users\nisch\Downloads\bank+marketing\bank\Original_data\bank.csv", sep=';')
print(df1.shape[1], df2.shape[1], df3.shape[1], df4.shape[1])
# Group files by their columns
# bank-additional-full & bank-additional likely have same columns
# bank-full & bank likely have same columns

# Combine files with matching columns
if list(df1.columns) == list(df2.columns):
    combined_additional = pd.concat([df1, df2], ignore_index=True).drop_duplicates()
    print(f"Combined additional: {len(combined_additional)} rows")

if list(df3.columns) == list(df4.columns):
    combined_bank = pd.concat([df3, df4], ignore_index=True).drop_duplicates()
    print(f"Combined bank: {len(combined_bank)} rows")

# Save separately if columns are different
combined_additional.to_csv(r"C:\Users\nisch\Downloads\bank+marketing\bank\Combined_data\combined_additional.csv", sep=';', index=False)
combined_bank.to_csv(r"C:\Users\nisch\Downloads\bank+marketing\bank\Combined_data\combined_bank.csv", sep=';', index=False)
```

```python
import os
import shutil

input_files = ['Combined_data\\combined_additional.csv', 'Combined_data\\combined_bank.csv']
out_dir = "final_datasets"
output_files = []

if out_dir and not os.path.exists(out_dir):
    os.makedirs(out_dir)
# Ensure output directories exist
import os
for file in input_files:
    file_name = os.path.split(file)[1]
    out_path = os.path.join(out_dir, file_name)
    output_files.append(out_path)
    shutil.copy(file, out_path)

print(input_files, output_files)


# Convert ; to , for each file
import csv
for input_file, output_file in zip(input_files, output_files):
    with open(input_file, 'r', newline='', encoding='utf-8') as infile, \
         open(output_file, 'w', newline='', encoding='utf-8') as outfile:
        reader = csv.reader(infile, delimiter=';')
        writer = csv.writer(outfile, delimiter=',')
        for row in reader:
            writer.writerow(row)
```
